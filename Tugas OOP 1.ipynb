```python
Tulis program Python untuk membuat kelas apapun dengan atribut instance-nya
```

```python
# Parent class 
class BatuSedimen:
    def __init__(self, nama, warna, tekstur):
        self.nama = nama        
        self.warna = warna      
        self.tekstur = tekstur  

    def intro(self):
        print(f"Tahukah kamu ada banyak jenis batuan sedimen, salah satunya {self.nama}?")

    def info(self):
        print(f"{self.nama} berwarna {self.warna} dan memiliki tekstur {self.tekstur}.")


# Child class 1
class Konglomerat(BatuSedimen):
    def formation(self):
        print(f"{self.nama} adalah batuan sedimen dengan fragmen bulat.")


# Child class 2
class Breksi(BatuSedimen):
    def formation(self):
        print(f"{self.nama} adalah batuan sedimen dengan fragmen menyudut.")


# Membuat objek dengan atribut instance
batu1 = Konglomerat("Konglomerat", "abu-abu", "kasar")
batu2 = Breksi("Breksi", "coklat kemerahan", "kasar")

# Memanggil method
batu1.intro()
batu1.info()
batu1.formation()

batu2.intro()
batu2.info()
batu2.formation()
```
